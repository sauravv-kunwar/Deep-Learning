
# Regularization in Deep Learning

## 1. What is Regularization?

Regularization is a technique used in Machine Learning and Deep Learning to prevent **overfitting** and improve the generalization ability of a model.

Overfitting occurs when a neural network learns the training data too well, including noise and unnecessary patterns. As a result, the model performs very well on training data but poorly on unseen or test data.

Regularization adds a penalty to the model's loss function to discourage overly complex models and prevent the weights from becoming excessively large.

### Why is Regularization Needed?

- Prevents overfitting in neural networks.
- Improves the model's ability to generalize to unseen data.
- Controls the magnitude of model weights.
- Reduces model complexity.
- Helps improve performance on test data.

### How Does Regularization Work?

Normally, a neural network tries to minimize its original loss function.

With regularization, an additional penalty term is added to the loss function.

**Original Loss Function:**

$$
L = L_{\text{original}}
$$

**Regularized Loss Function:**

$$
L_{\text{total}} = L_{\text{original}} + \lambda R(W)
$$

Where:

- $L_{\text{total}}$ = Total loss function.
- $L_{\text{original}}$ = Original loss (e.g., MSE or Binary Cross-Entropy).
- $\lambda$ = Regularization parameter.
- $R(W)$ = Regularization penalty based on model weights.

The regularization parameter $\lambda$ controls the strength of regularization.

- Small $\lambda$: Weak regularization.
- Large $\lambda$: Strong regularization.
- $\lambda = 0$: No regularization.

There are two commonly used regularization techniques:

1. L1 Regularization (Lasso).
2. L2 Regularization (Ridge).

---

## 2. L1 Regularization in ANN

L1 Regularization is a technique that adds the **absolute values of the model weights** to the original loss function.

It penalizes large weights and encourages some weights to become exactly zero.

### Mathematical Formula

The L1 regularized loss function is:

$$
L_{\text{total}} =
L_{\text{original}} +
\lambda \sum_{i=1}^{n}|w_i|
$$

Where:

- $L_{\text{original}}$ = Original loss function.
- $\lambda$ = Regularization parameter.
- $w_i$ = Individual model weights.
- $n$ = Total number of weights.

### How Does L1 Regularization Work?

1. Initially, the neural network calculates its original loss.
2. L1 regularization calculates the sum of the absolute values of all weights.
3. The penalty term is multiplied by the regularization parameter $\lambda$.
4. This penalty is added to the original loss.
5. During backpropagation, the model updates its weights to minimize the total loss.

L1 regularization encourages the model to reduce unnecessary weights and can force some weights to become exactly zero.

### Characteristics of L1 Regularization

- Encourages sparse weight matrices.
- Can make some weights exactly zero.
- Performs automatic feature selection in suitable models.
- Helps reduce model complexity.
- Useful when many input features are irrelevant.

### Advantages of L1 Regularization

- Removes less important features by driving their weights toward zero.
- Produces sparse models.
- Can improve interpretability.
- Helps prevent overfitting.

### Disadvantages of L1 Regularization

- May remove useful features if regularization is too strong.
- Can be unstable when features are highly correlated.
- Optimization can be more challenging because the absolute value function is not differentiable at zero.

---

## 3. L2 Regularization in ANN

L2 Regularization is a technique that adds the **sum of the squares of model weights** to the original loss function.

It discourages the model from assigning excessively large values to its weights.

L2 regularization is also commonly known as **Ridge Regularization** or Weight Decay in many neural network implementations.

### Mathematical Formula

The L2 regularized loss function is:

$$
L_{\text{total}} =
L_{\text{original}} +
\lambda \sum_{i=1}^{n}w_i^2
$$

Where:

- $L_{\text{original}}$ = Original loss function.
- $\lambda$ = Regularization parameter.
- $w_i$ = Individual model weights.
- $n$ = Total number of weights.

### How Does L2 Regularization Work?

1. The neural network performs forward propagation and calculates the original loss.
2. L2 regularization calculates the sum of the squared weights.
3. The penalty term is multiplied by the regularization parameter $\lambda$.
4. The penalty is added to the original loss.
5. During backpropagation, the network updates its weights to minimize the total loss.

L2 regularization continuously penalizes large weights, encouraging them to become smaller rather than forcing them to become exactly zero.

### Characteristics of L2 Regularization

- Penalizes large weights more strongly because weights are squared.
- Encourages smaller weight values.
- Generally does not make weights exactly zero.
- Produces smoother and more stable models.
- Helps reduce overfitting.

### Advantages of L2 Regularization

- Prevents weights from becoming excessively large.
- Improves model generalization.
- Works well when most features contribute to the prediction.
- Produces more stable models.
- Helps reduce sensitivity to individual training samples.

### Disadvantages of L2 Regularization

- Does not usually eliminate irrelevant features completely.
- Too much regularization can cause underfitting.
- Choosing an appropriate regularization parameter is important.

---

## 4. Difference Between L1 and L2 Regularization

Both L1 and L2 regularization help prevent overfitting, but they penalize model weights differently.

| L1 Regularization | L2 Regularization |
|---|---|
| Uses absolute values of weights. | Uses squared values of weights. |
| Penalty is $\lambda \sum |w_i|$. | Penalty is $\lambda \sum w_i^2$. |
| Encourages sparse weights. | Encourages smaller weights. |
| Can make weights exactly zero. | Usually does not make weights exactly zero. |
| Performs feature selection. | Retains most features with reduced weight values. |
| Also known as Lasso. | Also known as Ridge. |
| Useful when many features are irrelevant. | Useful when most features are relevant. |

---

## 5. Weight Decay in ANN

Weight Decay is a regularization technique used in neural networks to prevent overfitting by gradually reducing the magnitude of model weights during training.

It discourages the network from learning excessively large weights, helping the model become simpler and generalize better to unseen data.

Weight decay is closely related to L2 regularization.

### How Does Weight Decay Work?

During the training process, weight decay modifies the weight update rule so that weights gradually move toward zero.

In standard gradient descent, the weight update rule is:

$$
w_{\text{new}} = w_{\text{old}} -
\eta \frac{\partial L}{\partial w}
$$

With L2 regularization, the update becomes:

$$
w_{\text{new}} =
w_{\text{old}} -
\eta \left(
\frac{\partial L}{\partial w} + 2\lambda w
\right)
$$

Where:

- $w$ = Model weight.
- $\eta$ = Learning rate.
- $\lambda$ = Weight decay coefficient.
- $\frac{\partial L}{\partial w}$ = Gradient of the original loss function.

The additional term reduces the magnitude of weights during training.

### Characteristics of Weight Decay

- Discourages excessively large weights.
- Helps prevent overfitting.
- Improves generalization.
- Keeps model parameters relatively small.
- Commonly used in deep learning optimizers.

### Weight Decay vs L2 Regularization

Weight decay and L2 regularization are closely related, but they are not always mathematically identical.

- In standard SGD, L2 regularization and coupled weight decay produce equivalent weight updates when the same scaling convention is used.
- In optimizers such as Adam, adding an L2 penalty to the loss is not generally equivalent to applying decoupled weight decay.
- AdamW implements decoupled weight decay, where weight decay is applied separately from the gradient-based update.

---

## 6. Comparison of Regularization Techniques

| Technique | Main Purpose | Effect on Weights |
|---|---|---|
| L1 Regularization | Prevent overfitting and perform feature selection. | Makes some weights exactly zero. |
| L2 Regularization | Prevent overfitting by controlling weight magnitude. | Makes weights smaller. |
| Weight Decay | Reduce the magnitude of weights during optimization. | Gradually shrinks weights toward zero. |
| Dropout | Prevent neurons from becoming overly dependent on each other. | Randomly disables neurons during training. |

---

## 7. Conclusion

Regularization is an important technique in deep learning that helps prevent overfitting and improves the generalization ability of neural networks.

- **L1 Regularization:** Adds the absolute values of weights to the loss function and encourages some weights to become zero.
- **L2 Regularization:** Adds the squared values of weights to the loss function and encourages smaller weight values.
- **Weight Decay:** Reduces the magnitude of weights during the optimization process.
- **Dropout:** Randomly disables a proportion of neurons during training to reduce overdependence between neurons.

The choice of regularization technique depends on the dataset, model architecture, and learning problem.


# Code Demo

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
import seaborn as sns
from mlxtend.plotting import plot_decision_regions

import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.optimizers import Adam
```

```python
X,y = make_moons(100,noise = 0.25,random_state = 2)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X[:,0], X[:,1], c=y)
plt.show()
```

```python
model1 = Sequential()

model1.add(Dense(128,input_dim=2, activation="relu"))
model1.add(Dense(128, activation="relu"))
model1.add(Dense(1,activation='sigmoid'))

model1.summary()
```

```python
adam = Adam(learning_rate=0.01)
model1.compile(loss='binary_crossentropy', optimizer=adam, metrics=['accuracy'])

history1 = model1.fit(X, y, epochs=2000, validation_split = 0.2,verbose=0)
```

```python
plot_decision_regions(X, y.astype('int'), clf=model1, legend=2)
plt.xlim(-2,3)
plt.ylim(-1.5,2)
plt.show()
```

```python
plt.plot(history1.history['loss'])
plt.plot(history1.history['val_loss'])
```

```python
model2 = Sequential()

model2.add(Dense(128,input_dim=2, activation="relu",kernel_regularizer=tensorflow.keras.regularizers.l2(0.003)))
model2.add(Dense(128, activation="relu",kernel_regularizer=tensorflow.keras.regularizers.l2(0.003)))
model2.add(Dense(1,activation='sigmoid'))

model2.summary()
```

```python
adam = Adam(learning_rate = 0.01)
model2.compile(loss = 'binary_crossentropy',optimizer = adam, metrics = ['accuracy'])

history2 = model2.fit(X,y, epochs = 2000, validation_split =0.2,verbose = 0)
```

```python
plot_decision_regions(X, y.astype('int'), clf=model2, legend=2)
plt.xlim(-2,3)
plt.ylim(-1.5,2)
plt.show()
```

```python
plt.plot(history2.history['loss'])
plt.plot(history2.history['val_loss'])
```

```python
model1_weight_layer1 = model1.get_weights()[0].reshape(256)
model2_weight_layer1 = model2.get_weights()[0].reshape(256)
```

```python
sns.boxplot(model1_weight_layer1)
```
