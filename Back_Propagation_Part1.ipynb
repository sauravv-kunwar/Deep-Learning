# Backpropagation in Artificial Neural Networks

## 1. Introduction

Backpropagation, also known as **Backward Propagation of Errors**, is a fundamental algorithm used to train Artificial Neural Networks (ANNs) and Deep Neural Networks (DNNs).

It is an algorithm that calculates the gradients of a loss function with respect to the weights and biases of a neural network using the chain rule of calculus.

These gradients are then used by an optimization algorithm, such as Gradient Descent or Adam, to update the network parameters and minimize the prediction error.

In simple terms, backpropagation allows a neural network to learn from its mistakes.

When a neural network makes a prediction, the predicted output is compared with the actual output using a loss function. The resulting error is then propagated backward through the network to calculate how much each weight and bias contributed to the error.

The network uses this information to adjust its parameters and improve future predictions.

### Why is Backpropagation Important?

A neural network may contain thousands or even millions of weights and biases. Manually adjusting these parameters is practically impossible.

Backpropagation provides an efficient mathematical approach to calculate the contribution of every parameter to the overall loss.

Its main purposes are:

1. Calculate gradients of the loss function with respect to weights and biases.
2. Determine how each parameter influences the prediction error.
3. Propagate error information from the output layer toward the input layer.
4. Provide gradients to an optimizer for updating model parameters.
5. Enable neural networks to learn complex patterns from training data.

**Important:** Backpropagation calculates gradients; the optimizer uses those gradients to update the parameters.

---

## 2. Intuition Behind Backpropagation

Imagine training a neural network to identify whether an image contains a cat.

Initially, the network has randomly initialized weights. It processes the image and predicts that it does not contain a cat.

But the actual label indicates that the image contains a cat.

The network has therefore made a mistake.

The training process needs to answer two questions:

- How wrong was the prediction?
- Which weights contributed to the incorrect prediction, and how should they change?

The loss function answers the first question by measuring the prediction error.

Backpropagation answers the second question by calculating gradients.

The optimizer then uses these gradients to update the weights.

This process is repeated across the training dataset until the network learns useful patterns.

### The Core Idea

Backpropagation does not simply send the same error backward to every neuron.

Instead, it calculates how sensitive the final loss is to each intermediate value and parameter using the chain rule.

This is what makes it possible to train networks with multiple hidden layers.

## 3. Overall Training Process

Training a neural network involves four major operations:

1. Forward Propagation
2. Loss Calculation
3. Backward Propagation
4. Parameter Update

The complete training cycle is:

```text
             Input Dataset (X)
                     |
                     v
          +----------------------+
          |  Forward Propagation |
          +----------------------+
                     |
                     v
             Predicted Output
                     |
                     v
          +----------------------+
          |    Loss Function     |
          +----------------------+
                     |
                     v
               Calculate Loss
                     |
                     v
          +----------------------+
          |   Backpropagation    |
          +----------------------+
                     |
                     v
              Calculate Gradients
                     |
                     v
          +----------------------+
          |      Optimizer       |
          +----------------------+
                     |
                     v
             Update Weights
               and Biases
                     |
                     v
              Repeat Training
```

The forward pass calculates the prediction, whereas the backward pass calculates the gradients required to improve the prediction.

## 4. Neural Network Architecture and Gradient Flow

Consider a neural network with an input layer, two hidden layers, and an output layer.

Each neuron is connected to neurons in the next layer through weights.

### Forward Propagation

During forward propagation, information moves from the input layer toward the output layer.

```text
Input Layer       Hidden Layer 1      Hidden Layer 2      Output Layer

    x1  --------->    h1  ----------->    h4  ---------
      \              /  \                 /             \
       \            /    \               /               \
    x2  --------->    h2  ----------->    h5  -----------> ŷ
       \            \    /               \               /
        \            \  /                 \             /
    x3  --------->    h3  ----------->    h6  ---------/
```

### Backward Propagation

After calculating the loss, gradients flow backward through the network.

```text
Input Layer       Hidden Layer 1      Hidden Layer 2      Output Layer

    x1  <---------    h1  <-----------    h4  <---------
      \              /  \                 /             \
       \            /    \               /               \
    x2  <---------    h2  <-----------    h5  <----------- Loss
       \            \    /               \               /
        \            \  /                 \             /
    x3  <---------    h3  <-----------    h6  <---------/
```

**Note:** The diagrams illustrate the direction of information and gradient flow. In actual backpropagation, gradients are calculated using derivatives and chain-rule relationships rather than simply copying the prediction error backward.

## 5. Mathematical Foundation

To understand backpropagation, we need to understand three fundamental mathematical concepts.

### 5.1 Weighted Sum

Every neuron receives inputs from the previous layer.

Each input is multiplied by its corresponding weight, and the results are added together with a bias.

The weighted sum is:

$$
z = w_1x_1+w_2x_2+\cdots+w_nx_n+b
$$

In vector notation:

$$
z=W^TX+b
$$

Where:

- X = Input vector
- W = Weight vector
- b = Bias
- z = Weighted sum or pre-activation value

### 5.2 Activation Function

The weighted sum is passed through an activation function.

$$
a=f(z)
$$

The activation function introduces non-linearity into the neural network, allowing it to learn complex relationships.

Common activation functions include:

**Sigmoid:**

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

Derivative:

$$
\sigma'(z)=\sigma(z)(1-\sigma(z))
$$

**ReLU:**

$$
f(z)=\max(0,z)
$$

Derivative:

$$
f'(z)=
\begin{cases}
1 & z>0\\
0 & z<0
\end{cases}
$$

**Tanh:**

$$
f(z)=\tanh(z)
$$

Derivative:

$$
f'(z)=1-\tanh^2(z)
$$

### 5.3 Chain Rule of Calculus

The chain rule is the mathematical foundation of backpropagation.

It is used to calculate the derivative of a composite function.

Suppose:

$$
y=f(u)
$$

And:

$$
u=g(x)
$$

Then:

$$
\frac{dy}{dx}
=
\frac{dy}{du}
\times
\frac{du}{dx}
$$

In a neural network, the loss depends on the output, the output depends on weights and activations, and those activations depend on previous layers.

Therefore, the chain rule allows us to calculate how each weight contributes to the final loss.

For example:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial \hat{y}}
\times
\frac{\partial \hat{y}}{\partial z}
\times
\frac{\partial z}{\partial w}
$$

This equation represents the basic mathematical mechanism of backpropagation.

## 6. Computational Graph of Backpropagation

A computational graph represents the mathematical operations performed by a neural network.

It helps us understand how values and gradients move through a network.

Consider a single neuron:

$$
z=wx+b
$$

$$
\hat{y}=\sigma(z)
$$

$$
L=\frac{1}{2}(y-\hat{y})^2
$$

Its computational graph is:

```text
       x          w
        \        /
         \      /
          MULTIPLY
              |
              v
              z
              |
              + <----- b
              |
              v
          SIGMOID
              |
              v
             ŷ
              |
              v
        LOSS FUNCTION
              |
              v
              L
```

During the forward pass, each operation calculates and stores intermediate values.

During the backward pass, derivatives are calculated in reverse order.

For example:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial \hat{y}}
\frac{\partial \hat{y}}{\partial z}
\frac{\partial z}{\partial w}
$$

The computational graph makes it easier to understand why backpropagation starts from the loss and works backward through the operations.

## 7. Forward Propagation in a Multi-Layer Network

Consider a neural network with one hidden layer and one output layer.

Let:

- X = Input data
- W1 = Input-to-hidden weights
- b1 = Hidden layer bias
- W2 = Hidden-to-output weights
- b2 = Output layer bias

### Hidden Layer

First, calculate the weighted sum:

$$
Z_1=W_1X+b_1
$$

Apply the activation function:

$$
A_1=f(Z_1)
$$

### Output Layer

Calculate the weighted sum:

$$
Z_2=W_2A_1+b_2
$$

Apply the output activation function:

$$
A_2=g(Z_2)
$$

Therefore, the final prediction is:

$$
\hat{Y}=A_2
$$

The predicted output is then passed to the loss function.

## 8. Loss Function

A loss function measures the difference between the predicted output and the actual output.

It provides a numerical measurement of the network's prediction error.

### Mean Squared Error (MSE)

MSE is commonly used for regression problems.

$$
L=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

Where:

- y = Actual output
- ŷ = Predicted output
- n = Number of observations

### Binary Cross-Entropy

Binary Cross-Entropy is commonly used for binary classification.

$$
L=-\frac{1}{n}\sum_{i=1}^{n}
[y_i\log(\hat{y}_i)+(1-y_i)\log(1-\hat{y}_i)]
$$

### Categorical Cross-Entropy

Categorical Cross-Entropy is commonly used for multiclass classification.

$$
L=-\sum_{i=1}^{C}y_i\log(\hat{y}_i)
$$

Where C represents the number of classes.

**Important:** The loss function measures the prediction error. Backpropagation calculates the gradients needed to determine how the parameters should change.

## 9. Backpropagation: Mathematical Derivation

Now let's derive backpropagation mathematically.

Consider a single neuron:

$$
z=wx+b
$$

$$
\hat{y}=\sigma(z)
$$

For simplicity, use the following loss function:

$$
L=\frac{1}{2}(y-\hat{y})^2
$$

Our objective is to calculate:

$$
\frac{\partial L}{\partial w}
$$

Using the chain rule:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial \hat{y}}
\times
\frac{\partial \hat{y}}{\partial z}
\times
\frac{\partial z}{\partial w}
$$

### Step 1: Differentiate the Loss Function

$$
L=\frac{1}{2}(y-\hat{y})^2
$$

Differentiating with respect to predicted output:

$$
\frac{\partial L}{\partial \hat{y}}
=
\hat{y}-y
$$

### Step 2: Differentiate the Activation Function

For sigmoid:

$$
\hat{y}=\sigma(z)
$$

Its derivative is:

$$
\frac{\partial \hat{y}}{\partial z}
=
\hat{y}(1-\hat{y})
$$

### Step 3: Differentiate the Weighted Sum

We know:

$$
z=wx+b
$$

Therefore:

$$
\frac{\partial z}{\partial w}=x
$$

### Step 4: Combine the Derivatives

Substituting all derivatives:

$$
\frac{\partial L}{\partial w}
=
(\hat{y}-y)
\times
\hat{y}(1-\hat{y})
\times x
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w}
=
(\hat{y}-y)\hat{y}(1-\hat{y})x
}
$$

Similarly, the derivative with respect to bias is:

$$
\boxed{
\frac{\partial L}{\partial b}
=
(\hat{y}-y)\hat{y}(1-\hat{y})
}
$$

These derivatives tell us how the weight and bias affect the loss.

## 10. Backpropagation Through Multiple Layers

The real power of backpropagation becomes clear when working with deep neural networks.

Consider a network with L layers.

For each layer, we define:

- W(l) = Weight matrix of layer l
- b(l) = Bias vector of layer l
- Z(l) = Pre-activation value
- A(l) = Activation output
- δ(l) = Error signal or gradient with respect to Z(l)

### Forward Pass

For each layer:

$$
Z^{(l)}=W^{(l)}A^{(l-1)}+b^{(l)}
$$

$$
A^{(l)}=f^{(l)}(Z^{(l)})
$$

For the first layer:

$$
A^{(0)}=X
$$

The final layer produces:

$$
\hat{Y}=A^{(L)}
$$

### Backward Pass

The backward pass starts from the output layer.

First, calculate the output layer error signal:

$$
\delta^{(L)}
=
\frac{\partial L}{\partial A^{(L)}}
\odot
f'^{(L)}(Z^{(L)})
$$

Then propagate the error to the previous layer:

$$
\delta^{(l)}
=
(W^{(l+1)})^T\delta^{(l+1)}
\odot f'^{(l)}(Z^{(l)})
$$

This process continues until the first hidden layer is reached.

Here, ⊙ represents element-wise multiplication.

### Calculate Weight Gradients

For each layer:

$$
\frac{\partial L}{\partial W^{(l)}}
=
\delta^{(l)}(A^{(l-1)})^T
$$

### Calculate Bias Gradients

$$
\frac{\partial L}{\partial b^{(l)}}
=
\delta^{(l)}
$$

For a mini-batch, the gradients are averaged over the batch.

These equations form the basis of backpropagation in deep neural networks.

## 11. Weight and Bias Updates

After calculating gradients, the network uses an optimizer to update its parameters.

The most basic optimization algorithm is Gradient Descent.

### Weight Update

$$
W_{new}
=
W_{old}
-
\eta\frac{\partial L}{\partial W}
$$

### Bias Update

$$
b_{new}
=
b_{old}
-
\eta\frac{\partial L}{\partial b}
$$

Where:

- η = Learning rate
- W = Weight matrix
- b = Bias vector
- ∂L/∂W = Weight gradient
- ∂L/∂b = Bias gradient

The negative sign indicates that the parameters are updated in the direction that reduces the loss locally.

### Role of Learning Rate

The learning rate determines the size of each parameter update.

**High Learning Rate:**
- Large parameter updates.
- May overshoot the minimum.
- Training can become unstable.

**Low Learning Rate:**
- Very small parameter updates.
- Training becomes slow.
- Requires more iterations.

**Appropriate Learning Rate:**
- More stable training.
- Suitable parameter updates.
- Better convergence.

## 12. Numerical Example of Backpropagation

Let's calculate backpropagation using a single neuron.

Assume:

| Parameter | Value |
|---|---:|
| Input (x) | 1 |
| Weight (w) | 0.5 |
| Bias (b) | 0 |
| Actual output (y) | 1 |
| Learning rate (η) | 0.1 |

Use sigmoid activation and the loss:

$$
L=\frac{1}{2}(y-\hat{y})^2
$$

### Step 1: Forward Propagation

Calculate weighted sum:

$$
z=wx+b
$$

$$
z=(0.5)(1)+0
$$

$$
z=0.5
$$

Apply sigmoid:

$$
\hat{y}=\frac{1}{1+e^{-0.5}}
$$

$$
\hat{y}\approx0.62246
$$

### Step 2: Calculate Loss

$$
L=\frac{1}{2}(1-0.62246)^2
$$

$$
L\approx0.07127
$$

### Step 3: Calculate Weight Gradient

Using:

$$
\frac{\partial L}{\partial w}
=
(\hat{y}-y)\hat{y}(1-\hat{y})x
$$

Substitute the values:

$$
=(0.62246-1)(0.62246)(1-0.62246)(1)
$$

$$
\frac{\partial L}{\partial w}
\approx-0.08872
$$

### Step 4: Calculate Bias Gradient

$$
\frac{\partial L}{\partial b}
=
(\hat{y}-y)\hat{y}(1-\hat{y})
$$

$$
\frac{\partial L}{\partial b}
\approx-0.08872
$$

### Step 5: Update Weight

$$
w_{new}=w-\eta\frac{\partial L}{\partial w}
$$

$$
w_{new}=0.5-(0.1)(-0.08872)
$$

$$
\boxed{w_{new}=0.508872}
$$

### Step 6: Update Bias

$$
b_{new}=b-\eta\frac{\partial L}{\partial b}
$$

$$
b_{new}=0-(0.1)(-0.08872)
$$

$$
\boxed{b_{new}=0.008872}
$$

The weight and bias have been updated based on the calculated gradients.

The next forward pass will use the updated parameters.

This is one complete training iteration.

## 13. Backpropagation and Different Optimizers

Backpropagation calculates gradients, but different optimizers can use those gradients in different ways.

### Gradient Descent

Updates parameters using the current gradient.

$$
W=W-\eta\nabla_W L
$$

### Stochastic Gradient Descent (SGD)

Updates parameters using the gradient calculated from one training example at a time.

### Mini-Batch Gradient Descent

Uses a small subset of training examples to calculate gradients before updating parameters.

### Momentum

Uses a moving accumulation of previous gradients to help accelerate updates in consistent directions.

### RMSprop

Adjusts the learning rate for individual parameters using a moving average of squared gradients.

### Adam Optimizer

Combines momentum and adaptive learning rates to provide efficient parameter updates.

Adam is widely used in modern deep learning applications.

## 14. Problems Associated with Backpropagation

### 14.1 Vanishing Gradient Problem

The vanishing gradient problem occurs when gradients become extremely small while propagating backward through many layers.

As a result:

- Earlier layers receive very small gradients.
- Weight updates become extremely small.
- Training becomes slow.
- Deep networks may struggle to learn long-range dependencies.

**Solutions:**

- ReLU and its variants.
- Xavier or He initialization.
- Batch Normalization where appropriate.
- Residual connections.

### 14.2 Exploding Gradient Problem

The exploding gradient problem occurs when gradients become excessively large during backward propagation.

As a result:

- Parameter updates become very large.
- Training becomes unstable.
- Loss may increase dramatically.
- Numerical problems such as NaN values may occur.

**Solutions:**

- Gradient clipping.
- Proper weight initialization.
- Suitable learning rate.
- Normalization techniques where appropriate.

### 14.3 Computational Cost

Large neural networks may contain millions or billions of parameters.

Calculating gradients for all these parameters requires considerable processing power and memory.

### 14.4 Saddle Points

A saddle point is a point where the gradient may be zero even though the point is not a local minimum or maximum.

This can slow down optimization.

## 15. Epoch, Batch and Iteration

These terms are commonly used when training neural networks.

**Epoch:** One complete pass through the entire training dataset.

**Batch:** A subset of training samples processed together before updating the parameters.

**Iteration:** One parameter update based on one batch.

### Example

Suppose:

- Training samples = 1,000
- Batch size = 100
- Number of epochs = 10

Iterations per epoch:

$$
\frac{1000}{100}=10
$$

Total iterations:

$$
10\times10=100
$$

Therefore, the network performs 100 parameter updates during training.

## 16. Advantages of Backpropagation

1. Efficiently calculates gradients using the chain rule.
2. Allows neural networks to learn weights and biases automatically.
3. Supports networks containing multiple hidden layers.
4. Works with a wide range of differentiable activation and loss functions.
5. Is applicable to classification, regression, computer vision, and NLP.
6. Can be combined with different optimization algorithms.

## 17. Disadvantages of Backpropagation

1. Requires differentiable operations or suitable subgradients.
2. Can require significant computational resources for large networks.
3. May suffer from vanishing and exploding gradients.
4. Performance depends on suitable hyperparameter selection.
5. Requires representative training data for effective learning.

## 19. Complete Backpropagation Algorithm

The overall backpropagation algorithm can be summarized as follows:

1. Initialize weights and biases.
2. Pass input data through the neural network.
3. Calculate the predicted output using forward propagation.
4. Calculate the loss using an appropriate loss function.
5. Start backward propagation from the output layer.
6. Calculate the output layer gradients.
7. Propagate gradients through hidden layers using the chain rule.
8. Calculate gradients for all weights and biases.
9. Update parameters using an optimizer.
10. Repeat the process for multiple batches and epochs.
11. Evaluate the model's performance.

---

## 20. Final Conclusion

Backpropagation is one of the most important algorithms in artificial neural networks and deep learning.

It enables neural networks to learn by calculating the gradients of the loss function with respect to their weights and biases.

It uses the chain rule of calculus to propagate gradients backward from the output layer toward the input layer.

These gradients are then used by optimization algorithms such as Gradient Descent and Adam to update the network parameters and minimize prediction error.

Without backpropagation or another suitable gradient-computation method, training conventional deep neural networks would be extremely difficult.

### Key Takeaway

$$
\boxed{
\text{Forward Pass}
\rightarrow
\text{Loss Calculation}
\rightarrow
\text{Backward Pass}
\rightarrow
\text{Gradient Calculation}
\rightarrow
\text{Parameter Update}
}
$$

- Forward Propagation = Make predictions.
- Loss Function = Measure errors.
- Backpropagation = Calculate gradients.
- Optimizer = Update weights and biases.
- Training = Repeat the process to improve predictions.
