
# Activation Functions in Deep Learning

## 1. What are Activation Functions?

An Activation Function is a mathematical function applied to the output of a neuron in an Artificial Neural Network (ANN). It determines whether a neuron should be activated and what output it should produce.

In a neural network, each neuron receives inputs, multiplies them by corresponding weights, adds a bias, and passes the result through an activation function.

The output of a neuron before applying the activation function is called **Weighted Sum** or **Pre-Activation Value**.

### Mathematical Representation

First, a neuron calculates the weighted sum:

$$
z = \sum_{i=1}^{n} w_i x_i + b
$$

Then, the activation function is applied:

$$
a = f(z)
$$

Where:

- $x_i$ = Input values.
- $w_i$ = Weights.
- $b$ = Bias.
- $z$ = Weighted sum.
- $f(z)$ = Activation function.
- $a$ = Output of the neuron.

### Why are Activation Functions Needed?

Activation functions introduce **non-linearity** into neural networks.

Without activation functions, a neural network containing multiple layers would behave like a single linear transformation, regardless of how many hidden layers it contains.

Activation functions allow neural networks to:

- Learn complex and non-linear relationships.
- Identify complicated patterns in data.
- Solve complex classification and regression problems.
- Improve the learning capability of deep neural networks.
- Transform neuron outputs into suitable ranges.

### Types of Activation Functions

Some commonly used activation functions are:

1. Sigmoid Activation Function.
2. Tanh Activation Function.
3. ReLU Activation Function.
4. Leaky ReLU Activation Function.
5. Softmax Activation Function.
6. ELU Activation Function.

In this section, we will mainly focus on Sigmoid, Tanh, and ReLU.

## 2. Sigmoid Activation Function

The Sigmoid activation function is a non-linear activation function that converts any real-valued input into a value between **0 and 1**.

It produces an S-shaped curve and is commonly used in the output layer of binary classification problems.

### Mathematical Formula

$$
f(x) = \frac{1}{1 + e^{-x}}
$$

Where:

- $x$ = Input value.
- $e$ = Euler's number (approximately 2.718).
- $f(x)$ = Output between 0 and 1.

### How Does Sigmoid Work?

The Sigmoid function takes any real-valued input and transforms it into a probability-like value between 0 and 1.

- If the input is a large positive number, the output approaches 1.
- If the input is a large negative number, the output approaches 0.
- If the input is 0, the output is exactly 0.5.

For example:

| Input (x) | Output (Sigmoid) |
|---|---|
| -10 | 0.000045 |
| -5 | 0.0067 |
| 0 | 0.5 |
| 5 | 0.9933 |
| 10 | 0.99995 |

### Characteristics of Sigmoid

- Produces an S-shaped curve.
- Output range is between 0 and 1.
- It is a smooth and differentiable function.
- It is not zero-centered.
- Commonly used for binary classification.
- Its gradient becomes very small for extremely positive or negative inputs.

### Advantages of Sigmoid

- Converts outputs into a probability-like range between 0 and 1.
- Useful for binary classification problems.
- Smooth and differentiable.
- Easy to understand and implement.
- Suitable for predicting the probability of a positive class.

### Disadvantages of Sigmoid

**1. Vanishing Gradient Problem**

For very large positive or negative inputs, the gradient of the Sigmoid function becomes extremely small.

As a result, during backpropagation, the gradients may become too small to effectively update the weights of earlier layers.

This slows down learning and makes training deep neural networks difficult.

**2. Not Zero-Centered**

The output of the Sigmoid function is always positive, ranging from 0 to 1.

This can make optimization less efficient because gradients are not centered around zero.

**3. Computationally Expensive**

It requires the calculation of an exponential function, which is relatively more computationally expensive than simpler activation functions such as ReLU.

### Derivative of Sigmoid

The derivative of the Sigmoid function is:

$$
f'(x) = f(x)(1-f(x))
$$

The maximum derivative is 0.25, which occurs when the input is 0.

For very large positive or negative inputs, the derivative approaches zero, contributing to the vanishing gradient problem.

### Applications of Sigmoid

- Binary classification problems.
- Output layer of binary classification neural networks.
- Logistic regression.
- Probability estimation.

## 3. Tanh Activation Function

Tanh (Hyperbolic Tangent) is a non-linear activation function that converts input values into a range between **-1 and +1**.

It produces an S-shaped curve similar to the Sigmoid function, but its output is zero-centered.

### Mathematical Formula

$$
f(x) = \tanh(x)
$$

It can also be represented as:

$$
f(x) = \frac{e^x-e^{-x}}{e^x+e^{-x}}
$$

Where:

- $x$ = Input value.
- $e$ = Euler's number.
- $f(x)$ = Output between -1 and +1.

### How Does Tanh Work?

The Tanh function transforms input values into a range between -1 and +1.

- If the input is a large positive number, the output approaches +1.
- If the input is a large negative number, the output approaches -1.
- If the input is 0, the output is exactly 0.

For example:

| Input (x) | Output (Tanh) |
|---|---|
| -10 | -1 |
| -5 | -0.9999 |
| 0 | 0 |
| 5 | 0.9999 |
| 10 | 1 |

### Characteristics of Tanh

- Produces an S-shaped curve.
- Output range is between -1 and +1.
- It is zero-centered.
- It is smooth and differentiable.
- Produces both positive and negative outputs.
- Can suffer from the vanishing gradient problem.

### Advantages of Tanh

- Zero-centered output helps make optimization more efficient than Sigmoid in many situations.
- Produces both positive and negative values.
- Useful for learning patterns involving positive and negative relationships.
- Smooth and differentiable.
- Often performs better than Sigmoid in hidden layers of older neural network architectures.

### Disadvantages of Tanh

**1. Vanishing Gradient Problem**

When the input becomes extremely positive or negative, the output approaches +1 or -1.

In these saturated regions, the gradient becomes very small, causing the vanishing gradient problem.

**2. Computationally Expensive**

It involves exponential calculations, making it more computationally expensive than ReLU.

**3. Saturation**

For very large positive or negative inputs, the output changes very little even when the input changes significantly.

This reduces the gradient and slows down the learning process.

### Derivative of Tanh

The derivative of the Tanh function is:

$$
f'(x) = 1-\tanh^2(x)
$$

The maximum derivative is 1, which occurs when the input is 0.

As the input moves toward positive or negative infinity, the derivative approaches zero.

### Applications of Tanh

- Hidden layers in traditional neural networks.
- Recurrent Neural Networks (RNNs).
- Long Short-Term Memory (LSTM) networks.
- Problems requiring zero-centered activation values.

## 4. ReLU Activation Function

ReLU (Rectified Linear Unit) is one of the most widely used activation functions in Deep Learning.

It converts all negative input values into zero while keeping positive input values unchanged.

Unlike Sigmoid and Tanh, ReLU does not saturate for positive input values.

### Mathematical Formula

$$
f(x) = \max(0,x)
$$

It can also be represented as:

$$
f(x) =
\begin{cases}
0, & x \leq 0 \\
x, & x > 0
\end{cases}
$$

### How Does ReLU Work?

The ReLU function follows a simple rule:

- If the input is negative, the output is 0.
- If the input is positive, the output is equal to the input.
- If the input is 0, the output is 0.

For example:

| Input (x) | Output (ReLU) |
|---|---|
| -10 | 0 |
| -5 | 0 |
| -1 | 0 |
| 0 | 0 |
| 5 | 5 |
| 10 | 10 |

### Characteristics of ReLU

- Output range is between 0 and positive infinity.
- Converts negative values into zero.
- Keeps positive values unchanged.
- Computationally efficient.
- Introduces non-linearity into neural networks.
- Produces sparse activations because negative inputs become zero.
- Does not suffer from vanishing gradients in its positive region.

### Advantages of ReLU

**1. Computationally Efficient**

ReLU is simple to calculate because it only checks whether the input is positive or negative.

**2. Reduces the Vanishing Gradient Problem**

For positive inputs, the derivative of ReLU is 1.

Therefore, it helps gradients flow through positive activation regions without shrinking during backpropagation.

**3. Faster Training**

ReLU often allows neural networks to train faster compared to Sigmoid and Tanh.

**4. Sparse Activation**

ReLU converts all negative inputs into zero, meaning some neurons remain inactive for particular inputs.

This produces sparse activations and can help the network learn useful representations.

### Disadvantages of ReLU

**1. Dying ReLU Problem**

The Dying ReLU problem occurs when a neuron continuously receives negative inputs and produces an output of zero.

For negative inputs, the gradient of ReLU is also zero.

As a result, the neuron may stop updating its weights and become permanently inactive.

**2. Not Zero-Centered**

ReLU produces only non-negative outputs, which means its output is not centered around zero.

**3. Unbounded Output**

Unlike Sigmoid and Tanh, ReLU has no upper limit for positive inputs.

Its output can grow indefinitely.

### Derivative of ReLU

The derivative of ReLU is:

$$
f'(x) =
\begin{cases}
0, & x < 0 \\
1, & x > 0
\end{cases}
$$

At $x=0$, the derivative is mathematically undefined. In practice, implementations choose a particular subgradient, commonly zero.

### Applications of ReLU

- Hidden layers of Artificial Neural Networks.
- Deep Neural Networks (DNNs).
- Convolutional Neural Networks (CNNs).
- Image classification.
- Object detection.
- Computer vision applications.

## 5. Comparison Between Sigmoid, Tanh and ReLU

| Characteristics | Sigmoid | Tanh | ReLU |
|---|---|---|---|
| Output Range | 0 to 1 | -1 to +1 | 0 to +∞ |
| Curve | S-shaped | S-shaped | Piecewise linear |
| Zero-Centered | No | Yes | No |
| Computational Cost | Higher | Higher | Low |
| Vanishing Gradient | Yes | Yes | Not in positive region |
| Negative Inputs | Output between 0 and 0.5 | Negative output | Output is 0 |
| Positive Inputs | Output between 0.5 and 1 | Positive output | Output equals input |
| Common Usage | Binary classification output | Hidden layers and RNNs | Hidden layers of deep networks |

## 6. Which Activation Function Should We Use?

The choice of an activation function depends on the neural network architecture and the type of problem being solved.

**1. Sigmoid**
- Mainly used in the output layer of binary classification problems.
- Useful when predicting the probability of a single positive class.

**2. Tanh**
- Can be used in hidden layers, especially in traditional neural networks and recurrent neural networks.
- Useful when zero-centered output is desirable.

**3. ReLU**
- Commonly used in hidden layers of deep neural networks.
- Preferred in many modern architectures because of its computational efficiency and ability to reduce the vanishing gradient problem in positive activation regions.

---

## 7. Conclusion

Activation functions are essential components of neural networks because they introduce non-linearity and allow models to learn complex patterns.

- **Sigmoid:** Converts input values into a range between 0 and 1. It is mainly used for binary classification.
- **Tanh:** Converts input values into a range between -1 and +1. It is zero-centered.
- **ReLU:** Converts negative inputs into zero while keeping positive inputs unchanged. It is widely used in hidden layers of deep neural networks.

ReLU is generally preferred for hidden layers in modern deep learning models, while Sigmoid is commonly used for binary classification output layers.
