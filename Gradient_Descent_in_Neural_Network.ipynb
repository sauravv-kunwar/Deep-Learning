
# Gradient Descent in Neural Networks

## 1. Introduction to Gradient Descent

Gradient Descent is an optimization algorithm used in Machine Learning and Deep Learning to minimize the loss (or cost) function of a model.

In neural networks, the main objective during training is to find the optimal values of weights and biases that minimize the difference between predicted outputs and actual outputs.

Initially, a neural network starts with randomly initialized weights and biases. During training, it makes predictions, calculates the loss, and updates its parameters to reduce that loss.

Gradient Descent helps the network learn by continuously adjusting its weights and biases in the direction that reduces the loss function.

### Why Do We Need Gradient Descent?

- To minimize the loss function.
- To optimize the weights and biases of a neural network.
- To improve prediction accuracy.
- To reduce the difference between actual and predicted values.
- To help the neural network learn patterns from training data.

### Basic Working Principle

Gradient Descent works by calculating the gradient (derivative) of the loss function with respect to the model parameters.

The gradient indicates:

- The direction in which the loss function increases.
- How steeply the loss function changes.
- How the parameters should be adjusted to decrease the loss.

Since our objective is to minimize the loss, we update the parameters in the opposite direction of the gradient.

### Gradient Descent Update Rule

The general weight update equation is:

$$
w_{\text{new}} = w_{\text{old}} -
\alpha \frac{\partial L}{\partial w}
$$

Similarly, the bias update equation is:

$$
b_{\text{new}} = b_{\text{old}} -
\alpha \frac{\partial L}{\partial b}
$$

Where:

- $w$ = Weight of the neural network.
- $b$ = Bias of the neural network.
- $L$ = Loss function.
- $\alpha$ = Learning rate.
- $\frac{\partial L}{\partial w}$ = Gradient of loss with respect to weight.
- $\frac{\partial L}{\partial b}$ = Gradient of loss with respect to bias.

### Role of Learning Rate

The learning rate determines how large a step the algorithm takes while updating the weights and biases.

- **Small Learning Rate:** Updates parameters slowly and may require more training iterations.
- **Large Learning Rate:** Updates parameters quickly but may overshoot the minimum or cause unstable training.
- **Appropriate Learning Rate:** Helps the model converge efficiently toward a minimum of the loss function.

---

## 2. How Gradient Descent Works in Neural Networks

Gradient Descent is performed as part of the neural network training process.

The process involves the following steps:

### Step 1: Initialize Weights and Biases

Initially, the weights and biases of the neural network are assigned values, usually using random initialization methods.

### Step 2: Forward Propagation

Input data is passed through the neural network to generate predictions.

Each neuron performs a weighted sum followed by an activation function.

### Step 3: Calculate Loss

The predicted output is compared with the actual output using a loss function.

For example, Mean Squared Error (MSE) can be used for regression problems.

$$
MSE = \frac{1}{N}\sum_{i=1}^{N}(y_i-\hat{y}_i)^2
$$

Where:

- $y_i$ = Actual value.
- $\hat{y}_i$ = Predicted value.
- $N$ = Number of training samples.

### Step 4: Backpropagation

Backpropagation calculates the gradients of the loss function with respect to each weight and bias using the chain rule of differentiation.

These gradients indicate how much each parameter contributes to the loss.

### Step 5: Update Weights and Biases

Gradient Descent uses the calculated gradients to update the parameters in the opposite direction of the gradient.

$$
w = w - \alpha \frac{\partial L}{\partial w}
$$

### Step 6: Repeat the Process

The process of forward propagation, loss calculation, backpropagation, and parameter updating is repeated until the model reaches a satisfactory loss value or the maximum number of training epochs is completed.

### Overall Training Workflow

1. Initialize weights and biases.
2. Perform forward propagation.
3. Calculate the loss.
4. Perform backpropagation.
5. Calculate gradients.
6. Update weights and biases.
7. Repeat until the model converges.

---

## 3. Types of Gradient Descent

Gradient Descent can be classified into three major types based on how much training data is used to calculate the gradient before updating the model parameters.

1. Batch Gradient Descent (BGD)
2. Stochastic Gradient Descent (SGD)
3. Mini-Batch Gradient Descent

All three methods aim to minimize the loss function, but they differ in how frequently they update the model parameters.

---

# 4. Batch Gradient Descent (BGD)

## Definition

Batch Gradient Descent is an optimization algorithm in which the entire training dataset is used to calculate the gradient of the loss function before updating the weights and biases.

In this method, the model processes all training samples, calculates the average loss, computes the gradients, and then performs one parameter update.

This means that the weights are updated only after the entire training dataset has been processed.

## How Batch Gradient Descent Works

Suppose a dataset contains 1,000 training samples.

In Batch Gradient Descent:

1. The model processes all 1,000 training samples.
2. It calculates the loss for every sample.
3. The average loss is calculated.
4. Gradients are calculated using all 1,000 samples.
5. Weights and biases are updated once.

This entire process is repeated for each epoch.

## Mathematical Representation

For a dataset containing $N$ training samples, the average loss is:

$$
J(w) = \frac{1}{N}\sum_{i=1}^{N}L_i(w)
$$

The gradient is:

$$
\nabla J(w) =
\frac{1}{N}\sum_{i=1}^{N}\nabla L_i(w)
$$

The weight update equation becomes:

$$
w_{\text{new}} = w_{\text{old}} -
\alpha \frac{1}{N}
\sum_{i=1}^{N}\nabla L_i(w)
$$

Where:

- $N$ = Total number of training samples.
- $L_i(w)$ = Loss for the $i^{th}$ training sample.
- $\nabla L_i(w)$ = Gradient of the loss for the $i^{th}$ sample.
- $\alpha$ = Learning rate.

## Advantages of Batch Gradient Descent

1. **Stable Updates:** Since the entire dataset is used to calculate gradients, parameter updates are relatively stable.
2. **Accurate Gradient Estimation:** It calculates the exact gradient of the empirical average loss.
3. **Smooth Convergence:** It generally follows a smoother path toward a minimum compared to stochastic gradient descent.
4. **Deterministic Updates:** For fixed parameters and data, it produces consistent updates.

## Disadvantages of Batch Gradient Descent

1. **High Memory Requirement:** Processing the entire dataset at once may require a large amount of memory.
2. **Slow Updates:** The model must process all training samples before updating its parameters.
3. **Not Suitable for Very Large Datasets:** Training becomes computationally expensive when the dataset contains millions of samples.
4. **Less Frequent Updates:** The model updates its weights only once per epoch.

## Example

Suppose we have 10,000 training samples.

If we use Batch Gradient Descent:

- All 10,000 samples are processed.
- One gradient is calculated using the entire dataset.
- The weights are updated once.

Therefore, one epoch requires only one parameter update.

# 5. Stochastic Gradient Descent (SGD)

## Definition

Stochastic Gradient Descent (SGD) is an optimization algorithm in which the model updates its weights and biases after processing each individual training sample.

Unlike Batch Gradient Descent, which uses the entire dataset to calculate gradients, SGD calculates the gradient using only one training sample at a time.

The term stochastic refers to randomness because training samples are generally shuffled before processing.

## How Stochastic Gradient Descent Works

Suppose a dataset contains 1,000 training samples.

In Stochastic Gradient Descent:

1. One training sample is selected.
2. Forward propagation is performed for that sample.
3. The loss is calculated.
4. Backpropagation calculates the gradient.
5. Weights and biases are updated immediately.
6. The next training sample is processed.
7. This process continues until all samples are processed.

Therefore, the model updates its parameters after every individual sample.

## Mathematical Representation

For an individual training sample, the weight update equation is:

$$
w_{\text{new}} = w_{\text{old}} -
\alpha \nabla L_i(w)
$$

Where:

- $L_i(w)$ = Loss for an individual training sample.
- $\nabla L_i(w)$ = Gradient calculated using that sample.
- $\alpha$ = Learning rate.

Unlike Batch Gradient Descent, SGD does not calculate the gradient using the entire dataset before updating the parameters.

## Advantages of Stochastic Gradient Descent

1. **Frequent Updates:** The model updates weights after every training sample.
2. **Less Memory Requirement:** It processes one sample at a time, requiring less memory for gradient computation.
3. **Faster Initial Learning:** It can start updating the model without waiting for the entire dataset to be processed.
4. **Suitable for Large Datasets:** It can be useful when working with very large or continuously arriving datasets.
5. **Can Escape Shallow Local Minima:** Randomness in updates can help the optimization process move out of some shallow local minima or saddle regions.

## Disadvantages of Stochastic Gradient Descent

1. **Noisy Updates:** Since gradients are calculated using individual samples, updates can fluctuate significantly.
2. **Unstable Convergence:** The loss may increase and decrease during training rather than decreasing smoothly.
3. **Sensitive to Learning Rate:** Choosing an appropriate learning rate is important for stable training.
4. **May Not Reach an Exact Minimum:** Updates can continue fluctuating around a minimum unless the learning rate is reduced appropriately.

## Example

Suppose we have 10,000 training samples.

If we use Stochastic Gradient Descent:

- The first sample is processed and weights are updated.
- The second sample is processed and weights are updated again.
- The process continues for all 10,000 samples.

Therefore, one epoch results in 10,000 parameter updates.

# 6. Mini-Batch Gradient Descent

## Definition

Mini-Batch Gradient Descent is an optimization algorithm that combines the advantages of Batch Gradient Descent and Stochastic Gradient Descent.

In this method, the entire training dataset is divided into smaller groups called mini-batches.

The model processes one mini-batch at a time and updates its weights and biases after processing each mini-batch.

It is one of the most commonly used approaches for training neural networks and deep learning models.

## How Mini-Batch Gradient Descent Works

Suppose we have a dataset containing 1,000 training samples and choose a batch size of 100.

The dataset is divided into 10 mini-batches, each containing 100 samples.

The process works as follows:

1. Divide the training dataset into smaller mini-batches.
2. Process the first mini-batch containing 100 samples.
3. Calculate the average loss for the mini-batch.
4. Perform backpropagation to calculate gradients.
5. Update weights and biases.
6. Process the next mini-batch.
7. Repeat until all mini-batches are processed.

Therefore, the model updates its parameters 10 times in one epoch.

## Mathematical Representation

Suppose each mini-batch contains $B$ training samples.

The average loss for a mini-batch is:

$$
J_B(w) = \frac{1}{B}\sum_{i=1}^{B}L_i(w)
$$

The gradient is:

$$
\nabla J_B(w) =
\frac{1}{B}\sum_{i=1}^{B}\nabla L_i(w)
$$

The weight update equation becomes:

$$
w_{\text{new}} = w_{\text{old}} -
\alpha \frac{1}{B}
\sum_{i=1}^{B}\nabla L_i(w)
$$

Where:

- $B$ = Number of samples in a mini-batch.
- $L_i(w)$ = Loss of an individual sample.
- $\alpha$ = Learning rate.

## Advantages of Mini-Batch Gradient Descent

1. **Balanced Performance:** It combines the stability of Batch Gradient Descent with the frequent updates of SGD.
2. **Efficient Computation:** It allows multiple samples to be processed together.
3. **Better Hardware Utilization:** Modern GPUs can efficiently process mini-batches through parallel computation.
4. **Moderate Memory Requirement:** It requires less memory than processing the entire dataset at once.
5. **Faster Training:** It generally provides an efficient balance between computational speed and convergence stability.
6. **Widely Used in Deep Learning:** Most neural networks are trained using mini-batch-based optimization.

## Disadvantages of Mini-Batch Gradient Descent

1. **Batch Size Selection:** Choosing an appropriate batch size can affect training performance.
2. **Memory Usage:** Larger mini-batches require more memory.
3. **Noisy Updates:** Gradients are still estimates of the full-dataset gradient, so some fluctuations may occur.
4. **Additional Hyperparameter:** Batch size must be selected and tuned depending on the model and available hardware.

## Example

Suppose we have 10,000 training samples and choose a batch size of 100.

Number of mini-batches:

$$
\text{Number of Batches} =
\frac{\text{Total Samples}}{\text{Batch Size}}
$$

$$
= \frac{10000}{100} = 100
$$

Therefore:

- Total training samples = 10,000.
- Batch size = 100.
- Number of mini-batches = 100.
- Parameter updates per epoch = 100.

# 7. Comparison: Batch vs Stochastic vs Mini-Batch Gradient Descent

The three methods differ mainly in the number of training samples used to calculate the gradient before updating the model parameters.

| Feature | Batch GD | Stochastic GD | Mini-Batch GD |
|---|---|---|---|
| Training samples used per update | Entire dataset | One sample | Small group of samples |
| Parameter updates per epoch | 1 | N | N/B |
| Speed of individual updates | Slow | Fast | Moderate |
| Memory requirement | High | Low | Moderate |
| Gradient stability | High | Low | Moderate |
| Convergence | Smooth | Noisy | Relatively stable |
| GPU utilization | Can be efficient, but memory intensive | Often less efficient | Highly efficient |
| Large datasets | Less suitable | Suitable | Highly suitable |

Where:

- $N$ = Total number of training samples.
- $B$ = Mini-batch size.

# 9. What is Batch Size?

Batch size refers to the number of training samples processed together before the model updates its weights and biases.

For example, if a dataset contains 10,000 samples:

- Batch size = 1: One sample is processed per update (SGD).
- Batch size = 32: 32 samples are processed per update.
- Batch size = 64: 64 samples are processed per update.
- Batch size = 128: 128 samples are processed per update.
- Batch size = 10,000: The entire dataset is processed per update (Batch GD).

### Effect of Batch Size

**Small Batch Size:**

- Requires less memory.
- Produces more frequent updates.
- Introduces more noise into gradient estimates.
- May provide useful optimization behavior.

**Large Batch Size:**

- Requires more memory.
- Produces fewer updates per epoch.
- Provides a more accurate estimate of the full-dataset gradient.
- Can make better use of parallel hardware.

The appropriate batch size depends on the dataset, model architecture, hardware, and optimization requirements.

# 10. What is an Epoch and an Iteration?

These two terms are important when understanding neural network training.

## Epoch

An epoch refers to one complete pass of the entire training dataset through the neural network.

For example, if a dataset contains 10,000 samples, one epoch means the model has processed all 10,000 samples.

## Iteration

An iteration refers to one parameter update performed by the optimization algorithm.

In mini-batch training, one iteration generally means processing one mini-batch and updating the model parameters once.

### Relationship Between Epoch and Iteration

The number of iterations per epoch is:

$$
\text{Iterations per Epoch} =
\left\lceil \frac{N}{B} \right\rceil
$$

Where:

- $N$ = Total number of training samples.
- $B$ = Batch size.
- $\lceil \ \rceil$ = Ceiling function, which rounds up to the next integer.

For example:

- Total samples = 10,000.
- Batch size = 100.
- Iterations per epoch = 100.

If the model is trained for 50 epochs:

$$
\text{Total Iterations} =
\text{Epochs} \times \text{Iterations per Epoch}
$$

$$
= 50 \times 100
$$

$$
= 5000
$$

Therefore, the model performs 5,000 parameter updates during training.

# 11. Which Gradient Descent Method is Best?

There is no single gradient descent method that is best for every situation.

However, Mini-Batch Gradient Descent is widely preferred for training modern neural networks because it provides a good balance between computational efficiency, memory usage, and gradient stability.

### When to Use Batch Gradient Descent

- When the dataset is relatively small.
- When stable and accurate gradient calculations are required.
- When sufficient memory is available.

### When to Use Stochastic Gradient Descent

- When processing individual samples is necessary.
- When working with streaming or continuously arriving data.
- When frequent parameter updates are desired.

### When to Use Mini-Batch Gradient Descent

- When training deep neural networks.
- When working with medium or large datasets.
- When using GPUs for parallel computation.
- When a balance between speed and stability is required.

---

# 12. Conclusion

Gradient Descent is a fundamental optimization algorithm used to train neural networks by minimizing the loss function and updating weights and biases.

There are three major types of Gradient Descent:

1. **Batch Gradient Descent:** Uses the entire training dataset to calculate gradients and updates parameters once per epoch.
2. **Stochastic Gradient Descent:** Uses one training sample at a time and updates parameters after every sample.
3. **Mini-Batch Gradient Descent:** Uses a small group of training samples to calculate gradients and updates parameters after each mini-batch.

Among these three approaches, Mini-Batch Gradient Descent is commonly used in modern deep learning because it offers a practical balance between computational efficiency, memory usage, and stable optimization.
