# Loss Functions in Deep Learning

## 1. Introduction

Deep learning models learn patterns from training data by adjusting the weights and biases of neural networks. Initially, the model may produce inaccurate predictions. During training, it compares its predictions with the actual target values and calculates the error.

A **Loss Function** is a mathematical function used to measure the difference between the actual value and the predicted value. It tells us how well or poorly a model is performing.

The main objective of training a deep learning model is to minimize the loss function so that the model can make more accurate predictions.

### Basic Training Process

1. **Input:** The model receives training data.
2. **Forward Propagation:** Input data passes through the neural network to generate predictions.
3. **Loss Calculation:** The loss function compares predicted values with actual values.
4. **Backpropagation:** Gradients of the loss are calculated with respect to model parameters.
5. **Optimization:** An optimizer updates the weights and biases to reduce the loss.
6. **Iteration:** This process continues over multiple batches and epochs until the model learns the patterns.

---

## 2. What is a Loss Function?

A **Loss Function** is a mathematical function that calculates the error between the actual output (ground truth) and the predicted output of a machine learning or deep learning model.

In simple terms, it tells us how far the model's prediction is from the correct answer.

Let:

- $y$ = Actual value
- $\hat{y}$ = Predicted value
- $L$ = Loss function

The general expression is:

$$
L(y,\hat{y}) = \text{Error between actual and predicted values}
$$

The objective of model training is:

$$
\boxed{\text{Minimize } L(y,\hat{y})}
$$

### Example

Suppose a model predicts the price of a house.

| Actual Price | Predicted Price | Absolute Error |
|---|---|---|
| Rs. 50 lakh | Rs. 48 lakh | Rs. 2 lakh |
| Rs. 60 lakh | Rs. 65 lakh | Rs. 5 lakh |
| Rs. 40 lakh | Rs. 39 lakh | Rs. 1 lakh |

The loss function calculates the prediction errors using a mathematical formula.

Different loss functions treat errors differently. Some penalize large errors more heavily, while others are less sensitive to outliers.

### Important Points

- Measures the difference between actual and predicted values.
- Provides a numerical objective for model training.
- Helps the model learn by minimizing prediction errors.
- The choice of loss function depends on the type of problem.
- A lower loss generally indicates better predictions according to that particular loss function.

### Note: Loss Function vs. Cost Function
While often used interchangeably, there is a technical difference between the two:
- **Loss Function:** Refers to the error calculation for a **single training example** (or instance). For example, $L(y, \hat{y})$.
- **Cost Function:** Refers to the average or sum of the loss functions over the **entire training dataset** (or a batch). For example, the Mean Squared Error (MSE) is a cost function because it aggregates individual losses over $n$ samples: $J(\theta) = \frac{1}{n}\sum_{i=1}^{n} L(y_i, \hat{y}_i)$.

---

## 3. Why are Loss Functions Required in Deep Learning?

A neural network contains many trainable parameters, such as weights and biases. It requires a mathematical objective to determine whether its predictions are improving or becoming worse.

Loss functions are important because they:

1. **Measure Prediction Error:** Quantify the difference between actual and predicted values.
2. **Guide Learning:** Provide the objective that the optimizer tries to minimize.
3. **Support Backpropagation:** Their derivatives are used to calculate gradients for updating model parameters.
4. **Help Compare Models:** Different models can be compared using the same loss function.
5. **Influence Learning Behaviour:** Different loss functions penalize different types of errors.
6. **Match the Problem:** Different loss functions are suitable for regression and classification tasks.

### Loss and Parameter Updates

Let the model parameters be represented by $\theta$.

The objective is to find the parameters that minimize the loss:

$$
\theta^* = \arg\min_{\theta} L(\theta)
$$

Gradient Descent updates the parameters using:

$$
\theta_{\text{new}} =
\theta_{\text{old}} -
\eta \frac{\partial L}{\partial \theta}
$$

Where:

- $\eta$ = Learning rate
- $\frac{\partial L}{\partial \theta}$ = Gradient of the loss with respect to model parameters.

**Remember:** The loss function defines what the model should optimize, while the optimizer determines how the model parameters are updated.

## 4. Types of Loss Functions

Loss functions are mainly categorized according to the type of machine learning problem.

### 4.1 Regression Loss Functions

# PART A: REGRESSION LOSS FUNCTIONS

## 5. Mean Absolute Error (MAE)

**Mean Absolute Error (MAE)** calculates the average of the absolute differences between actual and predicted values.

It measures the average magnitude of prediction errors without considering their direction.

### Formula

For $n$ observations:

$$
\boxed{
MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
}
$$

Where:

- $n$ = Total number of observations
- $y_i$ = Actual value
- $\hat{y}_i$ = Predicted value
- $|\cdot|$ = Absolute value

### Example

Suppose the actual values are:

$$
y = [10,20,30]
$$

Predicted values:

$$
\hat{y} = [12,18,33]
$$

Absolute errors:

$$
|10-12|=2
$$

$$
|20-18|=2
$$

$$
|30-33|=3
$$

Therefore:

$$
MAE = \frac{2+2+3}{3}
$$

$$
\boxed{MAE = 2.33}
$$

The model's average absolute prediction error is approximately 2.33 units.

### Advantages of MAE

1. Easy to understand and calculate.
2. Expressed in the same unit as the target variable.
3. Less sensitive to outliers than MSE.
4. Treats positive and negative errors equally.

### Disadvantages of MAE

1. Does not penalize large errors as strongly as MSE.
2. Its gradient is not defined exactly at zero.
3. May be less convenient for some optimization methods than a smooth loss.

### When to Use MAE?

MAE is suitable when we want to measure the average magnitude of prediction errors without giving extremely large errors disproportionately high importance.

---

## 6. Mean Squared Error (MSE)

**Mean Squared Error (MSE)** calculates the average of the squared differences between actual and predicted values.

Since errors are squared, larger errors receive significantly higher penalties than smaller errors.

### Formula

$$
\boxed{
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
}
$$

### Example

Using the same values:

Actual values:

$$
y = [10,20,30]
$$

Predicted values:

$$
\hat{y} = [12,18,33]
$$

Squared errors:

$$
(10-12)^2=4
$$

$$
(20-18)^2=4
$$

$$
(30-33)^2=9
$$

Therefore:

$$
MSE = \frac{4+4+9}{3}
$$

$$
\boxed{MSE = 5.67}
$$

### Why are Errors Squared?

Squaring the errors has two important effects:

1. Positive and negative errors cannot cancel each other.
2. Large errors receive disproportionately higher penalties.

For example:

- Error of 2 gives squared error of 4.
- Error of 10 gives squared error of 100.

### Advantages of MSE

1. Simple and widely used for regression.
2. Penalizes large prediction errors strongly.
3. Smooth and differentiable, making it suitable for gradient-based optimization.
4. Has useful mathematical properties in statistical modelling.

### Disadvantages of MSE

1. Highly sensitive to outliers.
2. The unit of MSE is the square of the target variable's unit.
3. A few large errors can dominate the overall loss.

### When to Use MSE?

MSE is suitable when large prediction errors should be penalized heavily and the dataset does not contain problematic extreme outliers.

---

## 7. Root Mean Squared Error (RMSE)

**Root Mean Squared Error (RMSE)** is the square root of Mean Squared Error.

It measures the typical size of prediction errors while penalizing large errors.

### Formula

$$
\boxed{
RMSE =
\sqrt{
\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
}
}
$$

Using the previous example:

$$
MSE = 5.67
$$

$$
RMSE = \sqrt{5.67}
$$

$$
\boxed{RMSE \approx 2.38}
$$

### Advantages

1. Penalizes large errors more than MAE.
2. Expressed in the same unit as the target variable.
3. Easier to interpret than MSE in many situations.

### Disadvantages

1. Sensitive to outliers.
2. A small number of large errors can increase RMSE considerably.

### When to Use RMSE?

RMSE is useful when we want to evaluate regression models and give greater importance to larger prediction errors.

---

## 8. Huber Loss

**Huber Loss** combines the properties of Mean Absolute Error (MAE) and Mean Squared Error (MSE).

- For small errors, it behaves like MSE.
- For large errors, it behaves approximately like MAE.

This makes it less sensitive to outliers than MSE while maintaining smooth optimization around zero.

### Formula

$$
L_\delta(y,\hat{y}) =
\begin{cases}
\frac{1}{2}(y-\hat{y})^2,
& \text{if } |y-\hat{y}|\leq \delta\\[6pt]
\delta\left(|y-\hat{y}|-\frac{1}{2}\delta\right),
& \text{otherwise}
\end{cases}
$$

Where:

- $y-\hat{y}$ = Prediction error
- $\delta$ = Threshold controlling the transition between squared and absolute error behaviour.

### Explanation

1. If the absolute error is small, Huber Loss uses a squared-error penalty.
2. If the absolute error exceeds $\delta$, it uses a linear penalty.
3. Therefore, large errors do not grow quadratically as they do in MSE.

### Advantages of Huber Loss

1. More robust to outliers than MSE.
2. Smooth around zero.
3. Combines useful properties of MAE and MSE.

### Disadvantages of Huber Loss

1. Requires selection of the threshold $\delta$.
2. Slightly more complex than MAE and MSE.
3. May not be suitable for every regression problem.

### When to Use Huber Loss?

Huber Loss is suitable for regression problems where the dataset may contain outliers, but smooth optimization is still desirable.

## 9. Binary Cross-Entropy (BCE) Loss

**Binary Cross-Entropy (BCE)** is a loss function used for binary classification problems, where the target has only two possible classes.

Examples:

- Spam / Not Spam
- Disease / No Disease
- Fraud / Not Fraud
- Yes / No

The model generally produces a probability between 0 and 1 using a **Sigmoid Activation Function**.

### Formula

For one observation:

$$
\boxed{
BCE = -[y\log(\hat{y})+(1-y)\log(1-\hat{y})]
}
$$

For $n$ observations:

$$
BCE = -\frac{1}{n}\sum_{i=1}^{n}
[y_i\log(\hat{y}_i)+(1-y_i)\log(1-\hat{y}_i)]
$$

Where:

- $y$ = Actual binary label (0 or 1).
- $\hat{y}$ = Predicted probability of class 1.
- $\log$ = Natural logarithm.

### Example 1: Correct Prediction

Suppose:

$$
y=1,\quad \hat{y}=0.9
$$

Then:

$$
BCE = -[1\log(0.9)+0\log(0.1)]
$$

$$
BCE = -\log(0.9)
$$

$$
\boxed{BCE \approx 0.105}
$$

The loss is small because the model assigned a high probability to the correct class.

### Example 2: Incorrect Prediction

Suppose:

$$
y=1,\quad \hat{y}=0.1
$$

Then:

$$
BCE = -\log(0.1)
$$

$$
\boxed{BCE \approx 2.303}
$$

The loss is much higher because the model assigned a low probability to the correct class.

### Why is BCE Useful?

BCE strongly penalizes confident incorrect predictions.

- If the actual class is 1, predicting a probability close to 0 produces a large loss.
- If the actual class is 0, predicting a probability close to 1 produces a large loss.
- Correct predictions with high confidence produce smaller losses.

### Advantages of BCE

1. Designed specifically for binary classification.
2. Works naturally with probability predictions.
3. Penalizes confident incorrect predictions.
4. Commonly used with a sigmoid output layer.
5. Provides useful gradients for model training.

### Disadvantages of BCE

1. Standard BCE is intended for binary targets.
2. Numerical implementations must handle probabilities near 0 and 1 carefully.
3. Can be affected by class imbalance.

### When to Use BCE?

Use Binary Cross-Entropy when solving a binary classification problem with a single output probability.
