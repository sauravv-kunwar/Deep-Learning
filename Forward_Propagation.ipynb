
# Forward Propagation | How a Neural Network Predicts Output?

## 1. Introduction

Forward propagation is the process through which a neural network takes input data, processes it through different layers, and produces a predicted output.

During this process, information flows in the forward direction, from the input layer through the hidden layers to the output layer.

**Main purpose:** To calculate the predicted output using weights, biases, and activation functions.

## 2. How Forward Propagation Works

Forward propagation involves the following steps:

1. **Input Layer:** Receives the input features from the dataset.
2. **Weights and Biases:** Each input is multiplied by its corresponding weight, and a bias is added.
3. **Weighted Sum:** The weighted inputs are added together to calculate the net input.
4. **Activation Function:** The weighted sum is passed through an activation function to introduce non-linearity.
5. **Hidden Layers:** The output of one layer becomes the input of the next layer.
6. **Output Layer:** Produces the final prediction of the neural network.

## 3. Mathematical Representation

The output of a neuron is calculated in two steps.

### Step 1: Calculate Weighted Sum

The weighted sum of inputs is calculated using:

$$
z = \sum_{i=1}^{n} w_i x_i + b
$$

Where:

- $x_i$ = Input values
- $w_i$ = Weights
- $b$ = Bias
- $z$ = Weighted sum

### Step 2: Apply Activation Function

The weighted sum is passed through an activation function:

$$
a = f(z)
$$

Where:

- $f$ = Activation function
- $a$ = Output of the neuron

The activation function introduces non-linearity, allowing neural networks to learn complex patterns.

## 4. Example of Forward Propagation

Consider a neuron with the following values:

- Input: $x_1 = 2,\ x_2 = 3$
- Weights: $w_1 = 0.5,\ w_2 = 0.4$
- Bias: $b = 0.2$

**Step 1: Calculate weighted sum**

$$
z = (x_1 \times w_1) + (x_2 \times w_2) + b
$$

$$
z = (2 \times 0.5) + (3 \times 0.4) + 0.2
$$

$$
z = 2.4
$$

**Step 2: Apply Sigmoid Activation Function**

The sigmoid function is:

$$
f(z) = \frac{1}{1 + e^{-z}}
$$

Substituting the value:

$$
a = \frac{1}{1 + e^{-2.4}}
$$

$$
a \approx 0.917
$$

Therefore, the final output of the neuron is approximately **0.917**.

## 5. Activation Functions Used in Forward Propagation

Some commonly used activation functions are:

| Activation Function | Purpose |
|---|---|
| Sigmoid | Converts output into a value between 0 and 1. |
| ReLU | Returns 0 for negative values and the same value for positive inputs. |
| Tanh | Converts output into a value between -1 and 1. |
| Softmax | Converts output scores into probabilities for multiple classes. |

## 6. Characteristics of Forward Propagation

- Information flows from the input layer to the output layer.
- It uses weights, biases, and activation functions.
- It calculates the predicted output of a neural network.
- It does not update weights and biases.
- It is used during both training and prediction.
- Its output is used to calculate the loss during model training.

## 7. Conclusion

Forward propagation is an essential process in a neural network that passes input data through different layers to generate predictions. It performs mathematical calculations using weights, biases, and activation functions. The predicted output is then compared with the actual output to calculate the loss, which is later minimized using backpropagation.

**Remember:**
- Forward Propagation → Generates predictions.
- Backpropagation → Calculates gradients and updates weights and biases to minimize prediction error.


```python
import numpy as np

# Define activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Let's create a simple 2-layer neural network (Input -> Hidden -> Output)
# Input shape: (1, 3) - 1 sample with 3 features
X = np.array([[0.5, 0.8, -0.2]])

# Layer 1 (Hidden Layer with 4 neurons)
# Weights matrix shape: (3, 4)
W1 = np.array([
    [0.1, 0.5, -0.3, 0.8],
    [0.2, -0.1, 0.4, 0.6],
    [-0.4, 0.9, 0.1, -0.2]
])
b1 = np.array([[0.1, 0.2, -0.1, 0.0]]) # Biases shape: (1, 4)

# Layer 2 (Output Layer with 1 neuron)
# Weights matrix shape: (4, 1)
W2 = np.array([
    [0.3],
    [-0.5],
    [0.7],
    [0.1]
])
b2 = np.array([[0.2]]) # Bias shape: (1, 1)

print("Input (X):", X)
```

### 2. Forward Propagation Steps

Now, we step through the calculations programmatically.

```python
# Step 1: Compute Linear Combination for Hidden Layer
Z1 = np.dot(X, W1) + b1
print("Linear output of Hidden Layer (Z1):", Z1)

# Step 2: Apply Activation Function (Sigmoid) for Hidden Layer
A1 = sigmoid(Z1)
print("Activated output of Hidden Layer (A1):", A1)

# Step 3: Compute Linear Combination for Output Layer
Z2 = np.dot(A1, W2) + b2
print("Linear output of Output Layer (Z2):", Z2)

# Step 4: Apply Activation Function for Output Layer (Final Prediction)
output = sigmoid(Z2)
print("Final Predicted Output (A2):", output)
```
