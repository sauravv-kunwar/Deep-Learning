```python
X = df.iloc[:,0:-1]
y = df.iloc[:,-1]
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2,random_state = 1)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()

model.add(Dense(7,activation = 'relu',input_dim = 7))
model.add(Dense(7,activation = 'relu'))

model.add(Dense(1,activation = 'linear'))
# activation funciton for output layer of regression problem is always linear
```

```python
model.summary()
```

```python
model.compile(loss = 'mean_squared_error',optimizer = 'Adam')
```

```python
history = model.fit(X_train_scaled,y_train,epochs = 100, validation_split = 0.2)
```

```python
y_pred = model.predict(X_test_scaled)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

```python
import matplotlib.pyplot as plt
plt.plot(history.history['loss'])
plt.plot(history.history['va;loss'])
```

### Process Overview

Here is a simple breakdown of the machine learning process implemented in this notebook:

1. **Data Loading:** Read the graduate admission dataset from a CSV file into a pandas DataFrame.
2. **Exploratory Data Analysis (EDA):** Inspected the shape, data types, and descriptive statistics of the dataset, and checked for duplicate rows.
3. **Feature-Target Split:** Separated the dataset into feature variables ($X$) and the target variable ($y$, representing the "Chance of Admit").
4. **Train-Test Split:** Divided the data into training (80%) and testing (20%) sets to validate the model performance.
5. **Feature Scaling:** Applied `MinMaxScaler` to scale all the feature values between 0 and 1, ensuring stable neural network training.
6. **Model Architecture:** Constructed a Sequential Artificial Neural Network (ANN) using Keras with two hidden dense layers (7 neurons each, ReLU activation) and a single output neuron (Linear activation).
7. **Model Compilation & Training:** Compiled the model using the Mean Squared Error (MSE) loss function and the Adam optimizer, then trained it for 100 epochs with a validation split.
8. **Predictions & Evaluation:** Generated predictions on the scaled test set and evaluated the accuracy using the $R^2$ Score metric.
9. **Loss Visualization:** Plotted the training and validation loss curves over the 100 epochs to inspect model learning behavior.
