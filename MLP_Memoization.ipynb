# Memoization in Machine Learning (MLP – Multi-Layer Perceptron)

## 1. Introduction

In Machine Learning, computational efficiency plays an important role, especially when working with large datasets and complex models.

During the training of neural networks, the same computations may sometimes be performed repeatedly. Performing these computations again and again can consume additional time and computational resources.

To overcome this issue, a technique called **Memoization** can be used.

Memoization is an optimization technique in which the results of previously performed computations are stored in memory so that they can be reused whenever the same computation is required again.

Instead of calculating the same result repeatedly, the system retrieves the previously stored result.

This reduces redundant computations and can improve computational efficiency.

In the context of a Multi-Layer Perceptron (MLP), memoization can be useful when certain intermediate computations are repeated and their inputs have not changed.

---

## 2. What is Memoization?

### Definition

Memoization is a programming and computational optimization technique that stores the results of expensive function calls and returns the stored result when the same inputs occur again.

In simple terms:

**Memoization means calculating something once, storing its result, and reusing that result whenever the same calculation is needed.**

For example, suppose a function takes considerable time to calculate the square of a number.

If we repeatedly calculate the square of 5, the function performs the same operation each time.

Without memoization:

- First call: Calculate 5 × 5 = 25.
- Second call: Calculate 5 × 5 = 25 again.
- Third call: Calculate 5 × 5 = 25 again.

With memoization:

- First call: Calculate 5 × 5 = 25 and store the result.
- Second call: Retrieve 25 from memory.
- Third call: Retrieve 25 from memory.

Therefore, repeated calculations are avoided.

## 3. Why is Memoization Important in Machine Learning?

Machine Learning models often perform computationally expensive operations during training and prediction.

For example:

- Matrix multiplication
- Activation function calculations
- Feature transformations
- Repeated mathematical operations
- Intermediate computations

When the same computation is required multiple times with unchanged inputs, performing it repeatedly wastes computational resources.

Memoization helps by storing and reusing previously calculated results.

### Importance of Memoization

1. **Reduces Redundant Computation:** Avoids performing the same calculation multiple times.

2. **Improves Computational Efficiency:** Previously calculated results can be retrieved directly.

3. **Saves Processing Time:** Reduces the time required for repeated computations.

4. **Reduces Computational Overhead:** Avoids unnecessary execution of expensive operations.

5. **Useful for Repeated Operations:** Particularly helpful when a function receives the same inputs multiple times.

6. **Improves Program Performance:** Can make certain Machine Learning workflows faster.

## 4. Understanding MLP (Multi-Layer Perceptron)

Before understanding memoization in MLP, it is important to understand how an MLP works.

A **Multi-Layer Perceptron (MLP)** is a type of feedforward artificial neural network consisting of multiple layers of interconnected neurons.

It generally consists of three types of layers:

### 1. Input Layer

The input layer receives the input features from the dataset.

For example, in a house price prediction model, the input features may include:

- Area of the house
- Number of bedrooms
- Number of bathrooms
- Location

### 2. Hidden Layers

Hidden layers perform mathematical computations on the input data.

Each neuron calculates a weighted sum of its inputs, adds a bias, and applies an activation function.

The mathematical representation is:

$$
z = Wx+b
$$

Where:

- $W$ = Weight matrix
- $x$ = Input vector
- $b$ = Bias
- $z$ = Weighted sum

The activation function is then applied:

$$
a=f(z)
$$

Where:

- $a$ = Output of the neuron
- $f$ = Activation function

### 3. Output Layer

The output layer produces the final prediction of the network.

For example:

- Classification: Predicting whether an email is spam or not spam.
- Regression: Predicting the price of a house.

## 5. Memoization in MLP

In a Multi-Layer Perceptron, forward propagation involves several mathematical operations.

For each layer, the network calculates:

1. Weighted sum.
2. Bias addition.
3. Activation function.
4. Output of the layer.

These computations are performed repeatedly during model training.

Memoization can optimize repeated computations by storing previously calculated results and reusing them when the same inputs and computational conditions occur.

### Example

Suppose an MLP needs to calculate the output of a particular function for the same input multiple times.

Let:

$$
f(x)=x^2+2x+1
$$

For an input value of 5:

$$
f(5)=5^2+2(5)+1
$$

$$
=25+10+1
$$

$$
=36
$$

If the same input is provided again, the model or program can retrieve the previously calculated result instead of calculating it again.

This is the basic idea behind memoization.

### Important Consideration

In ordinary MLP training, forward propagation and backpropagation are generally performed again when weights or inputs change.

Memoization is only safe when the inputs and all relevant parameters affecting the computation remain unchanged.

Therefore, memoization is not automatically beneficial for every MLP operation.

---

## 6. How Memoization Works

Memoization generally follows four steps:

### Step 1: Receive Input

A function receives an input value.

### Step 2: Check Stored Results

Before performing the computation, the system checks whether the result for that input already exists in memory.

### Step 3: Reuse or Calculate

- If the result exists, retrieve the stored result.
- If the result does not exist, perform the calculation.

### Step 4: Store the Result

After calculating the result, store it in memory for future use.

### Working Flow

Input → Check Cache → Is Result Available?

- **Yes:** Return the stored result.
- **No:** Perform computation → Store result → Return result.

---

## 7. Mathematical Representation of Memoization

Suppose we have a function:

$$
y=f(x)
$$

Without memoization, the function is executed every time it receives an input.

If the same input is provided $n$ times, the computation is repeated $n$ times.

With memoization:

$$
y=
\begin{cases}
\text{Cache}[x], & \text{if } x \text{ exists in cache}\\
f(x), & \text{otherwise}
\end{cases}
$$

After calculating the result, it is stored:

$$
\text{Cache}[x]=f(x)
$$

The next time the same input is received, the stored result is returned directly.

### Computational Complexity

Suppose a particular computation takes $T$ seconds.

If the same calculation is performed $n$ times:

**Without Memoization:**

$$
T_{\text{total}}=nT
$$

**With Memoization:**

Assuming one initial calculation and negligible cache lookup time:

$$
T_{\text{total}}\approx T+(n-1)T_c
$$

Where:

- $T$ = Time required for the original computation.
- $T_c$ = Time required to retrieve a cached result.
- $n$ = Number of repeated calls.

Since cache retrieval is generally faster than performing an expensive computation, memoization can significantly reduce execution time.

## 8. Python Implementation of Memoization

Let's understand memoization using a simple Python example.

### Example 1: Without Memoization

```python
def square(n):
    print("Calculating...")
    return n * n

print(square(5))
print(square(5))
print(square(5))
```

**Output:**

```text
Calculating...
25
Calculating...
25
Calculating...
25
```

Here, the function performs the same calculation three times.

### Example 2: With Memoization

```python
cache = {}

def square(n):

    if n in cache:
        print("Fetching from cache...")
        return cache[n]

    print("Calculating...")

    result = n * n

    cache[n] = result

    return result


print(square(5))
print(square(5))
print(square(5))
```

**Output:**

```text
Calculating...
25
Fetching from cache...
25
Fetching from cache...
25
```

**Explanation:**

- During the first call, the square of 5 is calculated.
- The result (25) is stored in the cache.
- During the second call, Python checks the cache.
- Since the result already exists, it retrieves 25 directly.
- The same process happens during the third call.

Therefore, repeated calculations are avoided.

## 9. Using Memoization with Python Decorators

Python provides a built-in decorator called `lru_cache` in the `functools` module.

It can be used to implement memoization easily.

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def square(n):

    print("Calculating...")

    return n * n


print(square(5))
print(square(5))
print(square(5))
```

**Explanation:**

- `lru_cache` stores previously calculated function results.
- `maxsize=None` allows the cache to grow without a specified maximum number of entries.
- When the same argument is passed again, the cached result is returned.
- The function does not need to repeat the original computation.

---

## 11. Disadvantages and Limitations of Memoization

Although memoization improves computational efficiency, it also has some limitations.

1. **Memory Consumption:** Cached results require additional memory.

2. **Limited Benefit for Unique Inputs:** If every input is different, memoization provides little or no benefit.

3. **Cache Invalidation:** If the underlying data or parameters change, previously stored results may become outdated.

4. **Additional Overhead:** Checking and maintaining a cache also requires computational resources.

5. **Not Suitable for Every Operation:** Some computations depend on changing model weights, inputs, or training states, making cached results invalid.

6. **Cache Management:** Large caches may require memory limits and eviction policies.

---

## 12. Difference Between Memoization and Memorization

Memoization and memorization are two different concepts.

| Basis | Memoization | Memorization |
|---|---|---|
| Meaning | Storing previously computed results. | Learning or remembering information. |
| Purpose | Improve computational efficiency. | Retain learned information. |
| Application | Programming and optimization. | Learning and memory. |
| Main benefit | Avoids repeated computations. | Helps retain previously learned information. |
| Example | Caching the result of a mathematical function. | An ML model learning patterns from training data. |

**Note:** In Machine Learning, memorization can also refer to a model learning specific training examples instead of generalizing well to unseen data. This is different from computational memoization.

````python
#Memoization Method

def fibb(n,d):
  if n in d:
    return d[n]

  else:
    d[n] = fibb(n-1,d) + fibb(n-2,d)
    return d[n]

````
